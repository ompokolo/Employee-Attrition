```python
import pandas as pd
```

```python
df = pd.read_excel("mobile_customers.xlsx")
df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
import numpy as np

```

```python
columns_to_drop = [
    'Unnamed: 0', 'username', 'name',
    'credit_card_number', 'credit_card_security_code'
]
df = df.drop(columns=columns_to_drop, errors='ignore')
```

```python
print(df)
```

```python
age_bins = [0, 29, 45, 60, np.inf]
age_labels = ['Under 30', '30-45', '46-60', 'Over 60']
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)
```

```python
salary_bins = [0, 49999, 100000, 200000, np.inf]
salary_labels = ['Low (<50k)', 'Medium (50k-100k)', 'High (100k-200k)', 'Ultra High (>200k)']
df['salary_group'] = pd.cut(df['salary'], bins=salary_bins, labels=salary_labels)
```

```python
print(df[['age', 'age_group', 'salary', 'salary_group']].head())
```

```python
# removing the columns
columns_to_drop = ['address', 'residence', 'current_location', 'credit_card_expire', 'age', 'salary']
df_anonymized = df.drop(columns=columns_to_drop, errors='ignore')
```

```python
# Mask identifying strings
df_anonymized['customer_id'] = df_anonymized['customer_id'].astype(str).apply(lambda x: f"ID-XXXX{x[-4:]}")
```

```python
df_anonymized['email'] = '[PROTECTED_EMAIL]'
df_anonymized.to_excel("anonymised_user_data.xlsx", index=False)
print("Anonymization complete! File saved as 'anonymised_user_data.xlsx'")
```

```python
columns_to_drop = [
    'Unnamed: 0', 'username', 'name', 'address', 'residence',
    'current_location', 'credit_card_number', 'credit_card_security_code'
]
df_clean = df.drop(columns=columns_to_drop, errors='ignore').copy()
```

```python
df_clean['customer_id'] = df_clean['customer_id'].astype(str).apply(lambda x: f"ID-XXXX{x[-4:]}")
df_clean['email'] = '[PROTECTED_EMAIL]'
```

```python
salary_bins = [0, 49999, 100000, 200000, np.inf]
salary_labels = ['Low (<50k)', 'Medium (50k-100k)', 'High (100k-200k)', 'Ultra High (>200k)']
df_clean['salary_group'] = pd.cut(df_clean['salary'], bins=salary_bins, labels=salary_labels)
df_clean = df_clean.drop(columns=['salary'])
```

```python
age_bins = [0, 29, 45, 60, np.inf]
age_labels = ['Under 30', '30-45', '46-60', 'Over 60']
df_clean['age_group'] = pd.cut(df_clean['age'], bins=age_bins, labels=age_labels)
df_clean = df_clean.drop(columns=['age'])
```

```python
print("--- PREVIEW OF ANONYMISED INSIGHTSPARK DATA ---")
print(df_clean.head())
```
