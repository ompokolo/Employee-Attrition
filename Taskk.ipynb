```python
import pandas as pd
import numpy as np
```

```python
dtype_df = pd.DataFrame({
   "columns":df.columns,
   "Data Type": df.dtypes.astype(str).values
})
dtype_df
```

```python
df.describe(include= "object")
```

```python
missing_marker = [
    "",
    "na",
    "NA",
    "N/A",
    "null",
    "NULL",
    "None",
    "-"
]
df.replace(
    missing_marker,
    np.nan,
    inplace=True
)
```

```python
missing_report = pd.DataFrame({
    "Null values":df.isnull().sum(),
    "Null percentage":(df.isnull().sum() / len(df) * 100).round(2)
})
missing_report = missing_report[
    missing_report["Null values"] > 0
    ].sort_values(
      "Null percentage", ascending = False
    )
missing_report
```

```python
print(df.duplicated().sum())
```

```python
categorical_columns = df.select_dtypes(include = "object")
categorical_columns
```

```python
for col in categorical_columns:
  print(col)
  print(df[col].value_counts(dropna=False))
```

```python
for col in categorical_columns:
  df[col] = df[col].str.title()
```

```python
df.isnull().sum()
```

```python
df.dropna()
```

```python
df.describe()
```

```python
df.columns
```

```python
apples_cash = df[(df['product_name'] == 'Apple') & (df['payment_method'] == 'Cash')]
```

```python
# Q1 Answer: Total count purchased
total_apples_purchased = apples_cash['quantity'].sum()
```

```python
# Q2 Answer: Total money spent
total_cash_spent = apples_cash['total_amount'].sum()
```

```python
bakershire_non_members = df[(df['store'] == 'Bakershire') & (df['customer_type'] == 'Non-Member')]
```

```python
# Q3 Answer: Total money spent
total_bakershire_spend = bakershire_non_members['total_amount'].sum()
```

```python
#results
print(f"1. Total apples purchased in cash: {total_apples_purchased}")
print(f"2. Total cash spent on these apples: ${total_cash_spent:,.2f}")
print(f"3. Total money spent at Bakershire by non-members: ${total_bakershire_spend:,.2f}")
```
