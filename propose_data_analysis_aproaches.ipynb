```python
import pandas as pd
import numpy as np
from google.colab import files
```

```python
df = pd.read_excel("/content/commonwealth_bank_social_media_dataset.xlsx")
```

```python
df_posts = pd.read_excel("commonwealth_bank_social_media_dataset.xlsx", sheet_name="Social_Media_Posts")

print("--- EXCEL DATA ANALYSIS RESULTS ---\n")
```

```python
print("1. MOST FREQUENT TOPICS:")
# Swap 'topic' out for the exact column name if your dictionary uses something like 'topic_name'
if 'topic' in df_posts.columns:
    print(df_posts['topic'].value_counts().head())
else:
    print("Column 'topic' not found. Check your Data_Dictionary sheet for exact names.")
print("-" * 50)
```

```python
# Analysis 2: Compare engagement across topics, formats, and post types
# =====================================================================
print("2. ENGAGEMENT COMPARISONS:")
# Assuming columns are named 'topic', 'format', 'post_type', and 'engagement_score' (or 'likes'/'shares')
# We will use standard groupings. Adjust column names based on your dictionary.
for group_col in ['topic', 'format', 'post_type']:
    if group_col in df_posts.columns:
        print(f"\nAverage engagement grouped by {group_col}:")
        # If you have an explicit 'engagement_rate' or 'likes' column, swap 'engagement' with that name
        engagement_col = [col for col in df_posts.columns if 'engage' in col or 'like' in col][0]
        print(df_posts.groupby(group_col)[engagement_col].mean())
print("-" * 50)
```

```python
# Analysis 3: Review positive, negative, neutral, and mixed sentiment
# =====================================================================
print("3. SENTIMENT DISTRIBUTION BREAKDOWN:")
if 'sentiment' in df_posts.columns:
    print(df_posts['sentiment'].value_counts())
    print("\nPercentage distribution:")
    print(df_posts['sentiment'].value_counts(normalize=True) * 100)
print("-" * 50)
```

```python
# Analysis 4: Find posts with high views but low engagement
# =====================================================================
print("4. HIGH VIEWS BUT LOW ENGAGEMENT POSTS (Top 5 Anomalies):")
# Assuming column names like 'views' or 'impressions' exist
views_col = [col for col in df_posts.columns if 'view' in col or 'impression' in col]
engage_col = [col for col in df_posts.columns if 'engage' in col or 'like' in col]

if views_col and engage_col:
    # Calculate a ratio: views divided by engagement (high ratio = high views, low interactions)
    df_posts['view_to_engage_ratio'] = df_posts[views_col[0]] / (df_posts[engage_col[0]] + 1)
    high_views_low_engage = df_posts.sort_values(by='view_to_engage_ratio', ascending=False).head(5)
    # Print identifying columns (e.g., post ID or text snippet)
    id_col = 'post_id' if 'post_id' in df_posts.columns else df_posts.columns[1]
    print(high_views_low_engage[[id_col, views_col[0], engage_col[0], 'view_to_engage_ratio']])
print("-" * 50)
```

```python
# Analysis 5: Summarize customer questions, complaints, praise, and feedback
# =====================================================================
print("5. CATEGORY / POST TYPE SUMMARY:")
if 'post_category' in df_posts.columns:
    print(df_posts['post_category'].value_counts())
elif 'feedback_type' in df_posts.columns:
    print(df_posts['feedback_type'].value_counts())
else:
    print("Reviewing textual context keys instead:")
    print(df_posts['post_type'].value_counts() if 'post_type' in df_posts.columns else "No category column found.")
print("\n--- ANALYSIS COMPLETE ---")
```
